# Echo App - HTMX v4

## What this app does

The echo app is a minimal FastHTML example. It serves a single route (`/`) that:

1. Accepts a query parameter `q` (e.g. `/?q=hello`)
2. Prints the full request dict to the server console
3. Returns the value of `q` as the response

Since this app uses no special htmx attributes (no `hx-target`, `hx-swap`, WebSockets, or SSE), migrating to htmx v4 only requires switching the headers with `htmx=False, htmx4=True`. No other code changes are needed.

```
from fasthtml.common import *
from fasthtml.jupyter import *
from hx4_patch.core import *

app = FastHTML(middleware=[cors_allow], htmx=False, htmx4=True)
rt = app.route

@rt
def index(d:dict, q:str=""):
    print(str(d)[:1000])
    return q

srv = JupyUvi(app)
```

```
srv.stop()
```
