```python
from fasthtml.common import *
from fasthtml.jupyter import *
```

```python
app,rt = fast_app(htmx=False, htmx4=True)

def Incrementer(start=0, btn_txt='Increment'):
    return Section(
        Button(btn_txt),
        Prev("me('output', m).value++;"),
        Output(start))

@rt
def index():
    return Titled('Surreal Incrementer',
        Incrementer(),
        Incrementer(5, 'Do it'))

srv = JupyUvi(app)
```
