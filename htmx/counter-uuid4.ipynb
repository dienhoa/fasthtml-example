# counter-uuid4

> htmx v4 version of `counter-uuid.py`

```
from fasthtml.common import *
from fasthtml.jupyter import *
from hx4_patch.core import *
```

```
from uuid import uuid4
from string import Formatter

class Component(FT):
    def __str__(self): return f"document.getElementById('{self.id}')"

fh_cfg['ft_cls'] = Component
fh_cfg['auto_id'] = True
```

```
app,rt = fast_app(htmx=False, htmx4=True)

def Incrementer(start=0, btn_txt='Increment'):
    return Section(
        Style('me button {margin: 6px;}'),
        out := Output(start),
        Button(btn_txt, hx_on_click=f"{out}.value++")
    )

@rt("/")
def get():
    return Titled('Uuid Incrementer',
        Incrementer(),
        Incrementer(5, 'Do it'))

srv = JupyUvi(app)
```
