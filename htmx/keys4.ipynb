# keys4 — Keyboard Shortcuts (htmx v4)

Migrated from `keys.py`. Nothing much changes from v2 to v4 — the `hx-trigger` syntax (including filters and `from:` modifiers) works identically. The only change is using `fast_app(htmx=False, htmx4=True)`.

```
from fasthtml.common import *
from fasthtml.jupyter import *
from hx4_patch.core import *
from asyncio import sleep
from random import randint
from datetime import datetime
import secrets

app,rt = fast_app(htmx=False, htmx4=True)
```

```
def keytrig(key, mods=None, **kw):
    keyup = '&&'.join(listify(mods)+[f"event.key==='{key}'"])
    return {'hx_trigger': f"click, keyup[{keyup}] from:body"} | kw

@rt
async def more(): return datetime.now()

@rt
def index():
    return Titled('Keyboard demo',
        Div(id='targ'),
        Button('Time', hx_get=more, hx_target='#targ', **keytrig('r'))
    )
```
